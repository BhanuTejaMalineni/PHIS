```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Define the path to your raw data folder
data_raw_path = "/Users/bhanutejamalineni/phis_project/data/raw"

# --- Load and explore the Second Dataset (the relevant one) ---
try:
    df_second = pd.read_csv(os.path.join(data_raw_path, 'Second Dataset.csv'))
    print("Second Dataset loaded successfully for detailed processing.")
    print("Shape:", df_second.shape)
    print("Columns:", df_second.columns.tolist())
    print("First 5 rows:\n", df_second.head())
    print("\nMissing values in Second Dataset:\n", df_second.isnull().sum())
except FileNotFoundError:
    print(f"Error: Second Dataset.csv not found at {os.path.join(data_raw_path, 'Second Dataset.csv')}")
    # You might want to exit or handle this error gracefully in a real script
    exit("Cannot proceed without the Second Dataset.")

# Define a function for light preprocessing specifically for the Second Dataset
def preprocess_second_dataset(df):
    print("\nPreprocessing Second Dataset...")

    # Define the column mapping based on actual column names in df_second
    # These are the columns from df_second that we want to standardize
    column_mapping = {
        'Timestamp': 'timestamp',
        'Heart_Rate': 'heart_rate',
        'Step_Count': 'steps',
        'Body_Temperature': 'body_temperature',
        'Blood_Pressure': 'blood_pressure', # Keep as string for now, parse later if needed
        'Blood_Oxygen': 'blood_oxygen',
        'Activity_Status': 'activity_status'
    }

    # Apply renaming
    df_processed = df.rename(columns=column_mapping)
    print(f"Renamed columns: {column_mapping}")

    # Convert timestamp to datetime
    if 'timestamp' in df_processed.columns:
        df_processed['timestamp'] = pd.to_datetime(df_processed['timestamp'])
        print("Converted 'timestamp' to datetime.")
    else:
        print("Warning: 'timestamp' column not found after renaming. Check column_mapping.")

    # Sort by timestamp
    if 'timestamp' in df_processed.columns:
        df_processed = df_processed.sort_values('timestamp').reset_index(drop=True)
        print("Sorted by timestamp.")

    # Select only the relevant columns we want to simulate
    # Ensure Athlete_ID is included if you want to simulate for multiple users
    desired_cols = [
        'timestamp',
        'Athlete_ID', # Keep this to differentiate data streams if you want to simulate multiple users
        'heart_rate',
        'steps',
        'body_temperature',
        'blood_pressure',
        'blood_oxygen',
        'activity_status'
    ]
    df_final = df_processed[[col for col in desired_cols if col in df_processed.columns]].copy()

    # Handle Blood_Pressure: It's a string like '133/78'. For now, we'll keep it as a string.
    # If you need to use it numerically for ML, you'd split it into systolic/diastolic.

    print(f"Processed columns: {df_final.columns.tolist()}")
    print("First 5 rows of processed data:\n", df_final.head())
    print("Missing values in final processed data:\n", df_final.isnull().sum())
    return df_final

# Apply preprocessing
df_simulation_source = preprocess_second_dataset(df_second)

# Let's visualize a bit to see the data stream for one athlete
if 'timestamp' in df_simulation_source.columns and 'heart_rate' in df_simulation_source.columns and 'Athlete_ID' in df_simulation_source.columns:
    # Pick a sample athlete for visualization
    sample_athlete_id = df_simulation_source['Athlete_ID'].iloc[0] # Get the first athlete ID
    df_sample_athlete = df_simulation_source[df_simulation_source['Athlete_ID'] == sample_athlete_id].copy()

    if not df_sample_athlete.empty:
        plt.figure(figsize=(15, 7))
        sns.lineplot(x='timestamp', y='heart_rate', data=df_sample_athlete)
        plt.title(f'Heart Rate Over Time for Athlete: {sample_athlete_id}')
        plt.xlabel('Time')
        plt.ylabel('Heart Rate (BPM)')
        plt.grid(True)
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(15, 7))
        sns.lineplot(x='timestamp', y='steps', data=df_sample_athlete)
        plt.title(f'Step Count Over Time for Athlete: {sample_athlete_id}')
        plt.xlabel('Time')
        plt.ylabel('Step Count')
        plt.grid(True)
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

    else:
        print(f"No data found for sample athlete ID: {sample_athlete_id}")
else:
    print("Visualization skipped: Required columns (timestamp, heart_rate, Athlete_ID) not found.")

# Save this cleaned DataFrame to be used by the simulation script
output_processed_dir = "/Users/bhanutejamalineni/phis_project/data/processed"
if not os.path.exists(output_processed_dir):
    os.makedirs(output_processed_dir)
df_simulation_source.to_csv(os.path.join(output_processed_dir, 'simulation_data.csv'), index=False)
print(f"\nProcessed data saved to {os.path.join(output_processed_dir, 'simulation_data.csv')}")
```
