```python
# phiss_project/app.py

from flask import Flask, request, jsonify
from flask_sqlalchemy import SQLAlchemy
from datetime import datetime
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

app = Flask(__name__)

# --- Database Configuration ---
# Get database credentials from environment variables
DB_USER = os.getenv('DB_USER', 'phis_user') # Default to phis_user if not set
DB_PASSWORD = os.getenv('DB_PASSWORD', '70403') # IMPORTANT: Change this!
DB_NAME = os.getenv('DB_NAME', 'phis_db')
DB_HOST = os.getenv('DB_HOST', 'localhost')
DB_PORT = os.getenv('DB_PORT', '5432')

app.config['SQLALCHEMY_DATABASE_URI'] = \
    f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
app.config['SQLALCHEMY_TRACK_MODIFICATIONS'] = False

db = SQLAlchemy(app)

# --- Database Model Definition ---
class HealthMetric(db.Model):
    __tablename__ = 'health_metrics' # Explicit table name

    id = db.Column(db.Integer, primary_key=True)
    athlete_id = db.Column(db.String(50), nullable=False)
    timestamp = db.Column(db.DateTime, default=datetime.utcnow, nullable=False)
    heart_rate = db.Column(db.Integer)
    steps = db.Column(db.Integer)
    body_temperature = db.Column(db.Float)
    blood_pressure_systolic = db.Column(db.Integer)
    blood_pressure_diastolic = db.Column(db.Integer)
    blood_oxygen = db.Column(db.Integer)
    activity_status = db.Column(db.String(100))
    # Add other fields as needed based on your simulation data

    def __repr__(self):
        return f"<HealthMetric {self.athlete_id} at {self.timestamp}>"

    def to_dict(self):
        return {
            "id": self.id,
            "athlete_id": self.athlete_id,
            "timestamp": self.timestamp.isoformat() if self.timestamp else None,
            "heart_rate": self.heart_rate,
            "steps": self.steps,
            "body_temperature": self.body_temperature,
            "blood_pressure_systolic": self.blood_pressure_systolic,
            "blood_pressure_diastolic": self.blood_pressure_diastolic,
            "blood_oxygen": self.blood_oxygen,
            "activity_status": self.activity_status
        }

# --- API Endpoints ---

@app.route('/')
def home():
    return "Proactive Health Insight System Backend is Running!"

@app.route('/health_metrics', methods=['POST'])
def receive_health_metric():
    data = request.get_json()
    if not data:
        return jsonify({"message": "Invalid JSON data"}), 400

    # Parse blood_pressure string "systolic/diastolic"
    blood_pressure_str = data.get('blood_pressure')
    systolic = None
    diastolic = None
    if blood_pressure_str and '/' in blood_pressure_str:
        try:
            parts = blood_pressure_str.split('/')
            systolic = int(parts[0])
            diastolic = int(parts[1])
        except ValueError:
            pass # Keep them None if parsing fails

    try:
        new_metric = HealthMetric(
            athlete_id=data.get('Athlete_ID'),
            timestamp=datetime.fromisoformat(data.get('timestamp')),
            heart_rate=data.get('heart_rate'),
            steps=data.get('steps'),
            body_temperature=data.get('body_temperature'),
            blood_pressure_systolic=systolic,
            blood_pressure_diastolic=diastolic,
            blood_oxygen=data.get('blood_oxygen'),
            activity_status=data.get('activity_status')
        )
        db.session.add(new_metric)
        db.session.commit()
        return jsonify({"message": "Health metric received and stored", "id": new_metric.id}), 201
    except Exception as e:
        db.session.rollback()
        return jsonify({"message": f"Error storing metric: {str(e)}"}), 500

@app.route('/health_metrics/<string:athlete_id>', methods=['GET'])
def get_health_metrics(athlete_id):
    metrics = HealthMetric.query.filter_by(athlete_id=athlete_id).order_by(HealthMetric.timestamp.desc()).all()
    if not metrics:
        return jsonify({"message": f"No metrics found for Athlete_ID: {athlete_id}"}), 404
    return jsonify([metric.to_dict() for metric in metrics]), 200

# --- Database Initialization (for development) ---
@app.cli.command('init-db')
def init_db_command():
    """Initializes the database tables."""
    with app.app_context():
        db.create_all()
    print('Initialized the database.')

if __name__ == '__main__':
    # This block is typically for development. For production, use a WSGI server.
    app.run(debug=True, port=5000)
```
